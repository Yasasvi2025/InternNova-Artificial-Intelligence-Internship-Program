**Step 1: Import Pandas and Load Dataset**

```python
import pandas as pd

# Load dataset
df = pd.read_csv("students_cleaned.csv")

# Display first five rows
print("First Five Rows:")
print(df.head())
```

**Step 2: Check Dataset Shape**

```python
print("Dataset Shape:")
print(df.shape)
```

**Step 3: Check Data Types**

```python
print("Dataset Information:")
df.info()

print("\nData Types:")
print(df.dtypes)
```

**Step 4: Identify Missing Values**

```python
print("Missing Values:")
print(df.isnull().sum())
```

**Step 5: Remove or Fill Missing Values**

```python
# Fill missing numerical values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Fill missing categorical values, if any
df["Department"] = df["Department"].fillna(
    df["Department"].mode()[0]
)

# Remove duplicate records
df = df.drop_duplicates()

print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

**Step 6: Identify Features and Target**

```python
# Identify Input Features and Output Target

print("INPUT (X) - Features:")
print("1. Age")
print("2. Department")

print("\nOUTPUT (y) - Target:")
print("1. Marks")
```

**Step 7: Separate X and y**

```python
# Select features
X = df[["Age", "Department"]]

# Select target
y = df["Marks"]

print("Features (X):")
print(X.head())

print("\nTarget (y):")
print(y.head())

print("\nFeature Shape:", X.shape)
print("Target Shape:", y.shape)
```
