**Step 1: Import Libraries and Load Dataset**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the cleaned dataset
df = pd.read_csv("students_cleaned.csv")

print("Student Dataset:")
print(df)

print("\nTotal Records:", len(df))
```

**Step 2: Bar Chart — Average Marks by Department**

```python
# Calculate average marks for each department
department_marks = df.groupby("Department")["Marks"].mean()

# Create bar chart
plt.figure(figsize=(8, 5))
department_marks.plot(kind="bar")

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

**Step 3: Line Chart — Student Marks**

```python
# Sort dataset by Student ID
sorted_df = df.sort_values("Student_ID")

# Create line chart
plt.figure(figsize=(10, 5))
plt.plot(sorted_df["Student_ID"], sorted_df["Marks"], marker="o")

plt.title("Student Marks")
plt.xlabel("Student ID")
plt.ylabel("Marks")
plt.grid(True)
plt.tight_layout()
plt.show()
```

**Step 4: Histogram — Distribution of Marks**

```python
# Create histogram
plt.figure(figsize=(8, 5))
plt.hist(df["Marks"], bins=8, edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

**Step 5: Scatter Plot — Age vs Marks**

```python
# Create scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df["Age"], df["Marks"])

plt.title("Relationship Between Age and Marks")
plt.xlabel("Age")
plt.ylabel("Marks")
plt.grid(True)
plt.tight_layout()
plt.show()
```
