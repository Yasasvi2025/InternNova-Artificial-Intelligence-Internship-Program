```python
# Task 1: NumPy Fundamentals
# Name: Nune Venkata Yasasvi

import numpy as np

# 1. Creating NumPy Arrays

array_1d = np.array([10, 20, 30, 40, 50])

array_2d = np.array([[1, 2, 3],
                     [4, 5, 6]])

print("1D Array:")
print(array_1d)

print("\n2D Array:")
print(array_2d)
```

```python
# 2. Array Indexing and Slicing

print("\nArray Indexing:")
print("Element at index 2:", array_1d[2])

print("\nArray Slicing:")
print("Elements from index 1 to 3:", array_1d[1:4])

print("2D Array Element:", array_2d[1, 2])

print("First Row:", array_2d[0, :])
```

```python
# 3. Array Shape and Dimensions

print("\nArray Shape and Dimensions:")

print("1D Array Shape:", array_1d.shape)
print("1D Array Dimensions:", array_1d.ndim)

print("2D Array Shape:", array_2d.shape)
print("2D Array Dimensions:", array_2d.ndim)
```

```python
# 4. Basic Mathematical Operations

numbers = np.array([2, 4, 6, 8])

print("\nMathematical Operations:")
print("Addition:", numbers + 2)
print("Subtraction:", numbers - 2)
print("Multiplication:", numbers * 2)
print("Division:", numbers / 2)
```

```python
# 5. Statistical Functions

print("\nStatistical Functions:")

print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
print("Sum:", np.sum(numbers))
```
