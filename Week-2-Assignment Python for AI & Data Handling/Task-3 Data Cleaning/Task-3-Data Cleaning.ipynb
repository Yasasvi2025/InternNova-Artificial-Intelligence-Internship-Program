```python
# Task 3: Data Cleaning
# Name: Nune Venkata Yasasvi

import pandas as pd

# 1. Load the Dataset
df = pd.read_csv("students_dirty.csv")

print("1. Original Dataset:")
print(df)
```

```python
# 2. Before Cleaning Dataset Analysis

print("\n2. Dataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:")
print(df.isnull().sum().sum())

print("\nDuplicate Records:")
print(df.duplicated().sum())
```

```python
# 3. Correct Inappropriate Data Types

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print("\n3. Data Types After Conversion:")
print(df.dtypes)
```

```python
# 4. Count Missing Values After Type Conversion

print("\n4. Missing Values After Type Conversion:")
print(df.isnull().sum())
```

```python
# 5. Fill Missing Values

# Fill missing Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Marks with mean
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Convert values to integer
df["Age"] = df["Age"].astype(int)
df["Marks"] = df["Marks"].round().astype(int)

print("\n5. Missing Values After Filling:")
print(df.isnull().sum())
```

```python
# 6. Identify Duplicate Records

print("\n6. Duplicate Records:")
print(df[df.duplicated()])
```

```python
# 7. Remove Duplicate Records

df = df.drop_duplicates()

print("\n7. Dataset After Removing Duplicates:")
print(df)
```

```python
# 8. Final Dataset Analysis

print("\n8. Final Dataset Information:")
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

print("\nFinal Data Types:")
print(df.dtypes)

print("\nRemaining Missing Values:")
print(df.isnull().sum().sum())

print("\nRemaining Duplicate Records:")
print(df.duplicated().sum())
```

```python
# 9. Display Cleaned Dataset

print("\n9. Final Cleaned Dataset:")
print(df)
```

```python
# 10. Save Cleaned Dataset

df.to_csv("students_cleaned.csv", index=False)

print("\nCleaned Dataset Saved Successfully!")
```
